# Customer Churn Analysis

This project focuses on cleaning and analyzing customer data to understand factors influencing churn.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("churn-bigml-20.csv")
sns.countplot(x='Churn', data=df)
plt.title("Customer Churn Distribution")
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python
sns.boxplot(x='Churn', y='Total day minutes', data=df)
plt.title("Churn vs Day Minutes")
plt.show()
```

```python
sns.countplot(x='International plan', hue='Churn', data=df)
plt.title("International Plan vs Churn")
plt.show()
```

## Conclusion

This project analyzed customer churn data by performing data cleaning and exploratory data analysis.

Key findings include:
Customers with higher usage are more likely to churn
International plan users show higher churn rates
Certain features strongly influence customer retention
